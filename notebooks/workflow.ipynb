
# Viro3D workflow

This notebook gives a concise overview of the Viro3D processing and retrieval workflow.

## Data download

The original Viro3D data are available from Zenodo:

https://zenodo.org/records/15622906

Required files:

- `viro3d_metadata.tar.gz`
- `colabfold_pdb.tar.gz`

Direct downloads:

- Metadata: https://zenodo.org/records/15622906/files/viro3d_metadata.tar.gz?download=1
- ColabFold structures: https://zenodo.org/records/15622906/files/colabfold_pdb.tar.gz?download=1

The workflow is divided into four main parts.



# Part 1. Dataset preparation

This part prepares the Viro3D dataset and extracts the Baculoviridae subset used in the study.

### Scripts

```text
scripts/dataset/
├── Viro3D_dataset.cpp
├── count_viro3d_family.cpp
├── extract_baculoviridae.cpp
└── match_baculoviridae_structures.cpp
```

### Main steps

**`Viro3D_dataset.cpp`**

Filters the original Viro3D metadata using:

```text
Peptide Category == protein
ColabFold pLDDT >= 80
Protein Length >= 50
Protein Length <= 2000
```

**`count_viro3d_family.cpp`**

Counts the number of retained proteins in each viral family.

**`extract_baculoviridae.cpp`**

Extracts Baculoviridae proteins from the filtered Viro3D dataset and generates the corresponding FASTA data.

**`match_baculoviridae_structures.cpp`**

Matches Baculoviridae proteins with their corresponding ColabFold structure files.



# Part 2. Query and reference selection

This part constructs the query set and reference set used for sequence- and structure-based retrieval.

### Scripts

```text
scripts/query_selection/
├── count_viro3d_names.cpp
├── normalize_viro3d_names.cpp
└── select_baculoviridae_queries.cpp
```

### Main steps

**`normalize_viro3d_names.cpp`**

Standardizes protein names so that proteins with equivalent annotations can be grouped consistently.

**`count_viro3d_names.cpp`**

Counts standardized protein-name groups and summarizes their distribution across Baculoviridae species.

**`select_baculoviridae_queries.cpp`**

Selects representative query proteins.

The final query set contains:

```text
12 protein groups × 4 proteins = 48 queries
```

The four proteins in each group are selected from different ICTV species.

The selected protein groups are:

```text
chitinase
dna polymerase
vp1054
me53
sod
gp37
ubiquitin
lef-1
lef-4
p49
p33
egt
```

The complete retained Baculoviridae protein collection is used as the reference/search database.



# Part 3. Sequence and structure retrieval

The same query proteins and reference proteins are searched using both sequence- and structure-based methods.

### Scripts

```text
scripts/search/
├── run_mmseqs2.sh
└── run_foldseek.sh
```

### MMseqs2

**`run_mmseqs2.sh`**

Performs sequence-based protein retrieval using MMseqs2.

### Foldseek

**`run_foldseek.sh`**

Performs structure-based protein retrieval using Foldseek.

Using the same query/reference protein sets makes the MMseqs2 and Foldseek results directly comparable.



# Part 4. Comparison and downstream analysis

This part compares the MMseqs2 and Foldseek retrieval results.

### Script

```text
scripts/comparison/
└── compare_mmseqs_foldseek.cpp
```

**`compare_mmseqs_foldseek.cpp`**

Integrates the sequence- and structure-based search results and classifies query–target pairs into categories such as:

```text
Both
Sequence_only
Structure_only
```

The resulting files are then used for downstream analysis and visualization, including:

- MMseqs2 / Foldseek hit-category statistics
- Sequence identity vs. structural similarity
- Foldseek-only TM-score vs. structural coverage
- Representative sequence/structure retrieval cases



## Workflow summary

```text
Viro3D metadata + ColabFold structures
                |
                v
        Dataset preparation
                |
                v
      Query/reference selection
                |
        +-------+-------+
        |               |
        v               v
     MMseqs2         Foldseek
        |               |
        +-------+-------+
                |
                v
          Result comparison
                |
                v
      Analysis and visualization
```
